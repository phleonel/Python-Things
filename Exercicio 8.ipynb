```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
dfpaises = pd.read_csv('paises.csv', delimiter=';')

dfpaises['Region'] = dfpaises['Region'].str.strip() # remove os espaços em branco
dfpaises['Country'] = dfpaises['Country'].str.strip()

america_norte = dfpaises[dfpaises['Region'] == 'NORTHERN AMERICA'] # true false, e depois usa desses true e false pra selecionar os valores no array

plt.plot(america_norte['Country'], america_norte['Birthrate'], marker='o', label='Natalidade')
plt.plot(america_norte['Country'], america_norte['Deathrate'], marker='o', label='Mortalidade')

plt.title('Natalidade x Mortalidade - América do Norte')
plt.xlabel('País')
plt.ylabel('Taxa (por 1000 habitantes)')
plt.xticks(rotation=20)
plt.legend()
plt.show() # gráfico de linhas
```

**2.**

```python
dfspace = pd.read_csv('space.csv', delimiter=';')

dfspace['Company Name'] = dfspace['Company Name'].str.strip()
dfspace['Status Mission'] = dfspace['Status Mission'].str.strip()

# top 5 empresas por missões de sucesso e de falha
sucesso = dfspace[dfspace['Status Mission'] == 'Success']['Company Name'].value_counts().head(5) # tudo encadeado
falha = dfspace[dfspace['Status Mission'] == 'Failure']['Company Name'].value_counts().head(5)

plt.figure(figsize=(12, 5))

# da esquerda
plt.subplot(1, 2, 1)
plt.bar(sucesso.index, sucesso.values)
plt.title('Top 5 empresas - missões de sucesso')
plt.xlabel('Empresa')
plt.ylabel('Quantidade de missões')
plt.xticks(rotation=20)

# da direita
plt.subplot(1, 2, 2)
plt.bar(falha.index, falha.values, color='tab:red')
plt.title('Top 5 empresas - missões de falha')
plt.xlabel('Empresa')
plt.ylabel('Quantidade de missões')
plt.xticks(rotation=20)

plt.tight_layout() # ajuste automático
plt.show() # gráfico de barras
```

**3.**

```python
# só as missões da Roscosmos
roscosmos = dfspace[dfspace['Company Name'] == 'Roscosmos']

certo = (roscosmos['Status Mission'] == 'Success').sum()
errado = (roscosmos['Status Mission'] != 'Success').sum()

plt.pie([certo, errado], labels=['Deram certo', 'Deram errado'], autopct='%1.1f%%', colors=['tab:green', 'tab:red'])
plt.title('Missões da Roscosmos')
plt.show() # gráfico de torta
```

**4.**

```python
falha = dfspace[dfspace['Status Mission'] == 'Failure']['Company Name'].value_counts().head(5)

plt.bar(falha.index, falha.values, color='#8E44AD')
plt.title('Top 5 empresas com mais missões "Failure"')
plt.xlabel('Empresa')
plt.ylabel('Quantidade de missões')
plt.xticks(rotation=20)
plt.show() # gráfico de barras
```

```python
america_latina = dfpaises[dfpaises['Region'] == 'LATIN AMER. & CARIB']

plt.scatter(america_latina['GDP ($ per capita)'],
            america_latina['Literacy (%)'],
            s=america_latina['Population'] / 100000, # pra não ficar ridiculamente grande
            alpha=0.6)

plt.title('Renda per capita x Alfabetização - América Latina')
plt.xlabel('PIB per capita (US$)')
plt.ylabel('Taxa de alfabetização (%)')
plt.show() # gráfico de dispersão (scatter)
```
