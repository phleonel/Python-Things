```python
import numpy as np
```

```python
# formado apenas por 1's
arr = np.ones(8)
print(arr)

# aleatórios dentre 0 a 9
arr2 = np.random.randint(0, 9, 8)
print(arr2)

# soma dos dois arrays
arr3 = arr + arr2
print(arr3)

# soma de todos os elementos do array resultante
soma_total = arr3.sum()
print("Soma total:", soma_total)

# reshape condicional
if soma_total >= 40:
    # mais linhas do que colunas -> 4 linhas x 2 colunas
    matriz = arr3.reshape(4, 2)
else:
    # mais colunas do que linhas -> 2 linhas x 4 colunas
    matriz = arr3.reshape(2, 4)

print(matriz)
```

```python
# matriz 2x2
matriz = np.zeros([2,2])

# adiciona um 1 em uma posição aleatória
linha = np.random.randint(0, matriz.shape[0])
coluna = np.random.randint(0, matriz.shape[1])

matriz[linha, coluna] = 1

# campo minado
perdeu = False
pos = set()  # guardar as casas que o rapaizinho já colocou
total = matriz.size - 1

for i in range(3):
    print(f"Jogada {i+1} de 3")

    while True:
        linha2 = int(input("Escolha uma linha (0 ou 1): "))
        coluna2 = int(input("Escolha uma coluna (0 ou 1): "))

        if (linha2, coluna2) in pos:
            print("Você já jogou nessa posição! Escolha outra.")
        else:
            break

    if matriz[linha2, coluna2] == 1:
        print("Game Over! :( Try Again!")
        perdeu = True
        break

    pos.add((linha2, coluna2))

if not perdeu and len(pos) == total:
    print("Congratulations! You beat the game! :)")

print(matriz)
```

```python
# cria uma matriz de tamanho qualquer
matriz = np.zeros([4, 3])
print(matriz)

# extrai linhas e colunas
linhas, colunas = matriz.shape
print("Linhas:", linhas)
print("Colunas:", colunas)

# multiplica
total = linhas * colunas
print("Total de elementos:", total)

# par ou ímpar
if total % 2 == 0:
    print("A matriz pode virar um vetor unidimensional com número PAR de elementos.")
else:
    print("A matriz pode virar um vetor unidimensional com número ÍMPAR de elementos.")
```

```python
np.random.seed(10)
matriz = np.random.randint(1, 51, size=(4, 4))
print("Matriz:")
print(matriz)

# média de cada linha e cada coluna
media_linhas = matriz.mean(axis=1)
media_colunas = matriz.mean(axis=0)
print("Média de cada linha:", media_linhas)
print("Média de cada coluna:", media_colunas)

# maior valor entre as médias
maior_media_linha = media_linhas.max()
maior_media_coluna = media_colunas.max()
print("Maior média entre as linhas:", maior_media_linha)
print("Maior média entre as colunas:", maior_media_coluna)

# quantidade de aparições de cada número '-'
valores, contagens = np.unique(matriz, return_counts=True)
print("Valores:", valores)
print("Contagens:", contagens)

# números que aparecem exatamente 2 vezes
valores_2x = valores[contagens == 2]
print("Números que aparecem 2 vezes:", valores_2x)
```

```python
# matriz 3x3 de zeros
zeros = np.zeros((3, 3))
print("Zeros:")
print(zeros)

# matriz 3x3 de uns
uns = np.ones((3, 3))
print("Uns:")
print(uns)

# soma elemento a elemento
soma = zeros + uns
print("Soma:")
print(soma)

# multiplica por um escalar (escolhi 5)
escalar = 5
resultado = soma * escalar
print("Resultado (soma * escalar):")
print(resultado)

# reshape para vetor 1D de 9 elementos
vetor = resultado.reshape(9)
print("Vetor 1D:")
print(vetor)
```

```python
# números de 10 a 30, de 2 em 2
arr = np.arange(10, 31, 2)
print("Array:", arr)

# maior valor e seu índice
maior = arr.max()
indice_maior = arr.argmax()
print("Maior valor:", maior, "| índice:", indice_maior)

# menor valor e seu índice
menor = arr.min()
indice_menor = arr.argmin()
print("Menor valor:", menor, "| índice:", indice_menor)

# soma de todos os elementos
soma = arr.sum()
print("Soma:", soma)
```

```python
np.random.seed(7)

# dois arrays de tamanho 5, números entre 1 e 10
arr1 = np.random.randint(1, 11, 5)
arr2 = np.random.randint(1, 11, 5)
print("arr1:", arr1)
print("arr2:", arr2)

# concatena os dois em um terceiro
arr3 = np.concatenate((arr1, arr2))
print("arr3:", arr3)

# transforma em matriz 2x5
matriz = arr3.reshape(2, 5)
print("Matriz:")
print(matriz)
```
