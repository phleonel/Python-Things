```python
import numpy as np
```

```python
dataset = np.loadtxt('space.csv', delimiter = ';', dtype = str, encoding='utf-8')

header = dataset[0] # guarda o título das colunas
body = dataset[1:] # guarda só as missões (tudo a partir da linha 1)

status_mission = body[:, -1] # última coluna
filter = status_mission == 'Success'

qtd_sucesso = filter.sum()
total = len(body)

porcentagem = (qtd_sucesso / total) * 100

print(f'Missões com sucesso: {qtd_sucesso} de {total}')
print(f'Porcentagem de missões que deram certo: {porcentagem:.2f}%')
```

**2-**

```python
cost = body[:, 6].astype(float)   # convertendo a coluna Cost de string pra num

# filtra só as missões com custo disponível (x > 0)
cost_disponivel = cost[cost > 0]

media_gastos = cost_disponivel.mean()

print(f'Missões com custo disponível: {len(cost_disponivel)} de {len(cost)}')
print(f'Média de gastos: {media_gastos:.2f} milhões de dólares')
```

```python
location = body[:, 2]
usa = np.char.endswith(location, 'USA') # só me fala True or False
m_usa = body[usa] # pega só os True da lista, printando as linhas da coluna 2 q tem USA

print(f'Total de missões realizadas pelos Estados Unidos: {len(m_usa)}')
```

**4-**

```python
company = body[:,1] # só mostra as linhas das companhias
filter = company == "SpaceX" # True e False só mostra as linhas que tem spacex
spacex = body[filter] # só as linhas que tem spacex

cost = spacex[:,6].astype(float) # filtra só a coluna 6 das linhas que tem spacex

mais_cara = cost.argmax() # índice do maior custo
a_mais_cara = spacex[mais_cara]

print(f'Missão mais cara da SpaceX: {a_mais_cara[4]}')
print(f'Custo: {a_mais_cara[6]} milhões de dólares')
print(mais_cara)
```

**5-**

```python
company = body[:, 1] # coluna Company Name

empresas, qtds = np.unique(company, return_counts=True)

for empresa, qtd in zip(empresas, qtds):
    print(f'{empresa}: {qtd} missões')
```

```python
status_rocket = body[:, 5] # só mostra as linhas da coluna 5
retired = status_rocket == 'StatusRetired'# True e False yara yara...

qtd_retired = retired.sum() # pega e soma a quantidade
total = len(body)
porcentagem = (qtd_retired / total) * 100

print(f'Foguetes StatusRetired: {qtd_retired} de {total}') # lógica parecidinha com a da questão 1
print(f'Porcentagem: {porcentagem:.2f}%')
```
