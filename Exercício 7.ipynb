```python
import pandas as pd
```

```python
dfpaises = pd.read_csv('paises.csv', delimiter=';')
```

```python
import numpy as np

group_oceania = dfpaises['Region'].str.contains('OCEANIA')
paises_oceania = dfpaises[group_oceania]

# quais países são da oceania
print(paises_oceania['Country'])
#.str.strip().tolist() pra ficar mais bonitinho
print()

# quantos países são da oceania
quantos_oceania = paises_oceania['Country'].count()
print(f'{quantos_oceania} são da Oceania')

```

**2.**

```python
maior_populacao = dfpaises['Population'].max()
pais_maior_populacao = dfpaises[dfpaises['Population'] == maior_populacao]

nome = pais_maior_populacao['Country'].values[0]
regiao = pais_maior_populacao['Region'].values[0]

print(f'País com maior população: {nome}')
print(f'Região: {regiao}')
```

**3.**

```python
group_region = dfpaises.groupby('Region')
media_alfabetizacao = group_region['Literacy (%)'].mean()

print(media_alfabetizacao)
```
