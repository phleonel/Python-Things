Questão 4

```python
pessoa = []
for i in range(3):
  nome = input(f'Nome da pessoa {i+1}: ')
  peso = float(input(f'Peso da pessoa {i+1}: '))

  pessoa.append({
      "nome": nome,
      "peso": peso
  })

mais_pesada = pessoa[0]
mais_leve = pessoa[0]

for pessoa in pessoa:
  if pessoa["peso"] > mais_pesada["peso"]:
    mais_pesada = pessoa
  if pessoa["peso"] < mais_leve["peso"]:
    mais_leve = pessoa

print(f'Pessoa mais pesada: {mais_pesada}')
print(f'Pessoa mais leve: {mais_leve}')


```

Questão 5

```python
n = int(input('Quantas pessoas vão ser cadastradas? '))

pessoas = []

for i in range(n):
    print(f"\nPessoa {i+1}")
    nome = input("Nome: ")
    idade = int(input("Idade: "))
    sexo = input("Sexo (M/F): ").strip().upper()

    pessoas.append({
        "nome": nome,
        "idade": idade,
        "sexo": sexo
    })

soma_idades = 0
for pessoa in pessoas:
    soma_idades += pessoa["idade"]

media_idade = soma_idades/n

mulheres_menores_20 = 0
for pessoa in pessoas:
    if pessoa["sexo"] == "F" and pessoa["idade"] < 20:
        mulheres_menores_20 += 1

print(f"\nMédia de idade do grupo: {media_idade:.2f} anos")
print(f"Mulheres com menos de 20 anos: {mulheres_menores_20}")
```

```python
produtos = []

for i in range(3):
  nome = input(f"Nome do produto {i+1}: ")
  preco = float(input(f"Preço do produto {i+1}: "))
  quantidade = int(input(f"Quantidade em estoque do produto {i+1}: "))
  produtos.append({
      "nome": nome,
      "preco": preco,
      "quantidade": quantidade

  })

for produto in produtos:
  total = produto["preco"] * produto["quantidade"]
  print(f'{produto["nome"]} - Valor total em estoque: R${total}')
```
