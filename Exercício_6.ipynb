```python
import pandas as pd
import numpy as np

seriesAno1 = pd.Series({'Java': 16.25, 'C': 16.04, 'Python': 9.85})
seriesAno2 = pd.Series({'C': 16.21, 'Python': 12.12, 'Java': 11.68})
```

2.

```python
total_ano1 = seriesAno1.sum()
total_ano2 = seriesAno2.sum()
print(f"Total Ano 1: {total_ano1:.2f}%")
print(f"Total Ano 2: {total_ano2:.2f}%")
```

3.

```python
crescimento = seriesAno2 - seriesAno1
print(crescimento)
```

4.

```python
cresceram = crescimento[crescimento > 0]
print(cresceram)
```

5.

```python
taxa_crescimento = (seriesAno2 - seriesAno1) / seriesAno1
projecao = seriesAno2 * (1 + taxa_crescimento) ** 2
mais_popular = projecao.nlargest(1)
print(mais_popular)
```

6.

```python
df = pd.DataFrame(
    index=['A', 'B', 'C', 'D', 'E'],
    columns=['W', 'X', 'Y', 'Z'],
    data=np.random.randint(1, 50, [5, 4])
)

media_x_menor_30 = df[df['X'] < 30]['X'].mean()
print(media_x_menor_30)
```

7.

```python
media_linha_D = df.loc['D'].mean()
soma_linha_E = df.iloc[4].sum()
print(media_linha_D, soma_linha_E)
```

8.

```python
sub = df.loc[['A', 'C', 'E'], ['X', 'Y']]
soma_por_linha = sub.sum(axis=1)
soma_por_coluna = sub.sum(axis=0)
print(soma_por_linha)
print(soma_por_coluna)
```
